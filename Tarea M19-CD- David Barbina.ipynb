```python
# Importe de librerias
import matplotlib.pyplot as plt
import pandas as pd
import pylab as pl
import numpy as np
from ipywidgets import interact, FloatSlider # Esta librería ayudará mucho a la hora de querer probar parámetros sin tener que gráficar múltiples veces para ajustar la curva
import os
os.chdir("/Users/David Barbina/Desktop/Archivos EBAC/Entregable mexico")

# Leer datos
df = pd.read_excel("Mexico GDP.xlsx")
x = np.array(df["Periodo"], dtype=float)
y = np.array(df["GDP"], dtype=float)

# Normalizamos Y para trabajar con la curva dada 
y_min, y_max = y.min(), y.max()
y_norm = (y - y_min) / (y_max - y_min)

# Definición del modelo logístico descrito en la práctica
def logistic(x, b1, b2):
    return 1 / (1 + np.exp(b1 * (x - b2)))

# Función que grafica los datos y la curva por parámetros
def plot_logistic(b1=-0.1, b2=1990):
    plt.figure(figsize=(8,5))
    plt.scatter(x, y_norm, color="blue", label="Datos normalizados")
    
    # Crear puntos ordenados para hacer una gráfca de una curva propia a modificar con los parámetros
    x_fit = np.linspace(x.min(), x.max(), 200) # la función linspace que nos hace una gráfica con 200 puntos, en primera instancia no tiene forma y su rango es el de las x
    y_fit = logistic(x_fit, b1, b2)
    
    plt.plot(x_fit, y_fit, color="red", linewidth=2, label=f"Modelo logístico\nbeta1={b1:.3f}, beta2={b2:.1f}") # Aquí se dibuja la curva, con los datos especificados, el label está editado para verse bien y que contenga los decimales adcuados para cada parámetro
    
    plt.xlabel("Año")
    plt.ylabel("PIB normalizado")
    plt.title("Ajuste manual del modelo logístico")
    plt.legend()
    plt.grid(True)
    plt.show()

# Controles deslizantes para ajustar los parámetros
interact(plot_logistic, # Aquí se muestra la manera de interactuar los parámetros, definiendo en primera instancia, sus valores iniciales, sus rangos, sus nivel de paso y la descripción a desplegar
    b1=FloatSlider(value=-0.1, min=-0.5, max=0.5, step=0.01, description='beta1'),
    b2=FloatSlider(value=1990, min=x.min(), max=x.max(), step=1, description='beta2'))

# Nota: Tengo entendido que hay maneras de optimizar los parámtros con otra u otras librerías, pero me limité
# a este método de interacción de parámetros para desarrollar una manera activa pero rápida de conseguirlo.
```

```python
df["Año norm"] = df["Periodo"] / df["Periodo"].max()
df["GDP norm"] = df["GDP"] / df["GDP"].max()

x = np.array(df["Año norm"], dtype=float)
y = np.array(df["GDP norm"], dtype=float)

interact(plot_logistic,
    b1=FloatSlider(value=50, min=-500, max=500, step=5, description='beta1'),
    b2=FloatSlider(value=0.5, min=0.5, max=1.5, step=0.01, description='beta2'))
```
